```python
import pandas as pd
import numpy as np

# Define the date range
start_date = '2003-01-01'
end_date = '2013-12-31'

# Generate a date range
dates = pd.date_range(start=start_date, end=end_date)

# Generate random data for each column
temp = np.random.uniform(5, 28, len(dates))
humidity = np.random.uniform(12, 95, len(dates))
od = np.random.uniform(100, 500, len(dates))
ds = np.random.uniform(0, 100, len(dates))
turbidity = np.random.uniform(0, 100, len(dates))

# Create a dictionary with the data
data = {
    'date': dates,
    'temp': temp,
    'humidity': humidity,
    'OD': od,
    'DS': ds,
    'turbidity': turbidity
}

# Create a pandas DataFrame
df = pd.DataFrame(data)

# Save the DataFrame to a CSV file
df.to_csv('environmental_data.csv', index=False)

print("CSV file 'environmental_data.csv' created successfully.")
```

```python
# Read the CSV file into a DataFrame
df = pd.read_csv('environmental_data.csv')

# Convert the 'date' column to datetime objects
df['date'] = pd.to_datetime(df['date'])

# Set the 'date' column as the index
df.set_index('date', inplace=True)

# Display descriptive statistics
display(df.describe())

# Display information about the DataFrame
display(df.info())
```
